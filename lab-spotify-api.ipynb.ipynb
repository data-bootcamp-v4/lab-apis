```python
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials
from getpass import getpass

CLIENT_ID = input("Spotify Client ID: ")
CLIENT_SECRET = getpass("Spotify Client Secret: ")

sp = spotipy.Spotify(
    auth_manager=SpotifyClientCredentials(
        client_id=CLIENT_ID,
        client_secret=CLIENT_SECRET
    )
)

print("Spotify client initialized!")
```

```python
artists = ["Lady Gaga", "The Weeknd", "Ed Sheeran"]

print("My favorite artists:")

for artist in artists:
    print(artist)
```

```python
sample_tracks = {
    "Lady Gaga": ["Bad Romance", "Poker Face", "Just Dance", "Shallow", "Born This Way"],
    "The Weeknd": ["Blinding Lights", "Starboy", "Save Your Tears", "The Hills", "Die For You"],
    "Ed Sheeran": ["Shape of You", "Perfect", "Thinking Out Loud", "Photograph", "Bad Habits"]
}

def get_top_tracks(artist):
    return sample_tracks.get(artist, [])[:5]

top_tracks_list = [get_top_tracks(artist) for artist in artists]

for artist, tracks in zip(artists, top_tracks_list):
    print(artist, ":", tracks)
```

```python
sample_related_artists = {
    "Lady Gaga": ["Katy Perry", "Rihanna", "Dua Lipa", "Madonna", "Ariana Grande"],
    "The Weeknd": ["Drake", "Frank Ocean", "Bruno Mars", "Post Malone", "SZA"],
    "Ed Sheeran": ["Shawn Mendes", "James Arthur", "Sam Smith", "Lewis Capaldi", "John Mayer"]
}

def find_related_artists(artist):
    return sample_related_artists.get(artist, [])[:5]

related_artists_list = [
    find_related_artists(artist) for artist in artists
]

for artist, related in zip(artists, related_artists_list):
    print(artist, ":", related)
```

```python
playlist = []

for artist in artists:
    playlist.extend(get_top_tracks(artist))

    related_artists = find_related_artists(artist)

    for related_artist in related_artists:
        playlist.extend(get_top_tracks(related_artist))

playlist = list(dict.fromkeys(playlist))

print("My Music Playlist:")
for number, song in enumerate(playlist, start=1):
    print(f"{number}. {song}")

print("Total songs:", len(playlist))
```
