```python
import pandas as pd
from bs4 import BeautifulSoup
import requests 
from IPython.display import Image
import warnings 
warnings.filterwarnings('ignore')
```

```python
url = 'https://swapi.dev/api/people/'
response = requests.get(url)
response.status_code
```

```python
data = response.json()
```

```python
characters = []
for index, i in enumerate(data['results'], start = 1): # helped as I wanted to add index start with one. 
    character_info = {'ID': index, 
                      'Name': i['name'],
                      'Gender': i['gender'],
                      'Birth Year': i['birth_year']
                     }
    characters.append(character_info)
df = pd.DataFrame(characters)
```

```python
df
```

```python
link = 'https://swapi.dev/api/films/'
response = requests.get(link)
response.status_code
```

```python
data_ = response.json()
```

```python
movies = []
for i in (data_['results']): 
    movie_info = {'Title': i['title'],
                  'Episode': i['episode_id'],
                  'Director': i['director'],
                  'Release': i['release_date'].split('-')[0] #split and keep the date. 
                     }
    movies.append(movie_info)
df2 = pd.DataFrame(movies)
```
