## Using the search method

```python
import requests
import base64

# Your Spotify app credentials
CLIENT_ID = "26b4b2177619406f8951cb297ea7ceb2"
CLIENT_SECRET = "038d36c7361b48859b0db05f531741a0"

# Step 1: Encode credentials in Base64
auth_str = f"{CLIENT_ID}:{CLIENT_SECRET}"
b64_auth = base64.b64encode(auth_str.encode()).decode()

# Step 2: Spotify Token URL
token_url = "https://accounts.spotify.com/api/token"

# Step 3: Prepare headers and data
headers = {
    "Authorization": f"Basic {b64_auth}",
    "Content-Type": "application/x-www-form-urlencoded"
}

data = {
    "grant_type": "client_credentials"
}

# Step 4: Request token
response = requests.post(token_url, headers=headers, data=data)

# Step 5: Get JSON response
result = response.json()
print(result)  # 🔍 Debug: see the full response

# Step 6: Extract the access token safely
token = result.get("access_token")

if token:
    print("\n✅ Your Access Token:")
    print(token)
else:
    print("\n❌ Failed to generate token. Check your credentials.")
```

```python
import requests

url = "https://api.spotify.com/v1/search"

headers = {
    "Authorization": f"Bearer {token}"
}

params = {
    "q": "AR Rahman",     # search keyword
    "type": "track",      # we want songs
    "limit": 10
}

response = requests.get(url, headers=headers, params=params)

info = response.json()

# Print song names
for track in info["tracks"]["items"]:
    print(track["name"], "-", track["artists"][0]["name"])
```

```python
import requests

def get_top_tracks(artist_name, token):
    headers = {"Authorization": f"Bearer {token}"}
    
    # 1️⃣ Search for the artist
    search_url = "https://api.spotify.com/v1/search"
    params = {"q": f"artist:{artist_name}", "type": "artist", "limit": 1}
    response = requests.get(search_url, headers=headers, params=params)
    data = response.json()
    
    items = data.get("artists", {}).get("items", [])
    if not items:
        print(f"Artist '{artist_name}' not found!")
        return []
    
    artist_id = items[0]["id"]
    
    # 2️⃣ Try multiple markets to get top tracks
    top_tracks_url = f"https://api.spotify.com/v1/artists/{artist_id}/top-tracks"
    markets = ["US", "GB", "IN"]  # ✅ Define BEFORE the loop
    tracks_data = []

    for market in markets:
        response = requests.get(top_tracks_url, headers=headers, params={"market": market})
        tracks_data = response.json().get("tracks", [])
        if tracks_data:  # Stop when we get results
            break
    
    if not tracks_data:
        print(f"No top tracks found for {artist_name} in any market!")
        return []

    # 3️⃣ Return first 5 track names
    return [track["name"] for track in tracks_data[:5]]
```

```python
# 1️⃣ Define the playlist ID first
playlist_id = "1qnQdgRvUHFGlY2VYnCe0O?si=1057ba4454404862"

# 2️⃣ Use it in the URL
playlist_url = f"https://api.spotify.com/v1/playlists/{playlist_id}"

# 3️⃣ Make request with token
headers = {"Authorization": f"Bearer {token}"}
response = requests.get(playlist_url, headers=headers)
playlist_data = response.json()

# 4️⃣ Print result
print(playlist_data)
```

### Getting Artists from a Playlist

To extract all the artists from the tracks in a playlist, you'd typically follow these steps:

1. Fetch the playlist's tracks.
2. Iterate through each track.
3. For each track, extract the associated artists.

```python
# List to store unique artist names
artists_list = []

for track_item in tracks['items']:
    track = track_item['track']
    for artist in track['artists']:
        artist_name = artist['name']
        if artist_name not in artists_list:  # This ensures each artist is added only once
            artists_list.append(artist_name)

print(artists_list)
```
