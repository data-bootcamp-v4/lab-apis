# Challenge 1: Pokémon Dataset & Analysis

## Goal

Build and analyze a Pokémon dataset using the API.

## Example (single request)

Below you can see how to get data for ONE Pokémon:

```python
import requests

url = "https://pokeapi.co/api/v2/pokemon/1"

response = requests.get(url)
data = response.json()

data
```

## Tasks

1. Using the example above, get data for Pokémon IDs from 1 to 151

2. Extract:
   - name
   - height
   - weight
   - base_experience
   - first type

3. Create a DataFrame

4. Add a new column:
   - weight_kg (divide weight by 10)

5. Answer:
   - Which Pokémon is the heaviest?
   - What is the average weight?
   - How many Pokémon are there per type?

6. Answer:
   - Which type has the highest average weight?

```python
"Using the example above, get data for Pokémon IDs from 1 to 151"
import pandas as pd
import time
from IPython.display import JSON
url = "https://pokeapi.co/api/v2/pokemon"

response = requests.get("https://pokeapi.co/api/v2/pokemon?limit=151&offset=0", timeout=10)
data = response.json()
display(data)

```

```python
"""Extract:
name
height
weight
base_experience
first type"""
pokemon_1to151=[]

for pokemon_id in range (1,152) :
    response = requests.get(f"https://pokeapi.co/api/v2/pokemon/{pokemon_id}", timeout=10)
    response.raise_for_status()
    data = response.json()
    pokemon_1to151.append({"id":data["id"], "name":data["name"], "height":data["height"], "weight": data["weight"], "base_experience": data ["base_experience"],"first_type": data["types"][0]["type"]["name"] 
                          })
    time.sleep(0.5)


df_pokemon = pd.DataFrame(pokemon_1to151)
display(len(pokemon_1to151))
```

```python
"""Create a DataFrame"""
df_pokemon = pd.DataFrame(pokemon_1to151)
display(df_pokemon)
```

```python
"""Add a new column:weight_kg (divide weight by 10)"""
df_pokemon["weight_kg"]=df_pokemon["weight"]/10



display(df_pokemon)
```

```python
"""Answer:"""

"""Which Pokémon is the heaviest?"""

heaviest = df_pokemon.loc[df_pokemon["weight_kg"].idxmax()]
heaviest
```

```python
"""What is the average weight?"""
average_weight=df_pokemon["weight_kg"].mean()
print(f"the average weight is {average_weight.round(2)} kg")
```

```python
"""How many Pokémon are there per type?"""
count_per_type=df_pokemon["first_type"].value_counts()
count_per_type
```

```python
"""Which type has the highest average weight?"""
highest_average_weight_per_type=df_pokemon.groupby("first_type")["weight_kg"].mean().idxmax()

average_weight_per_type=df_pokemon.groupby("first_type")["weight_kg"].mean()
average_weight_per_type["rock"]

print(f"{highest_average_weight_per_type} has the highest average weight, with {average_weight_per_type["rock"].round(2)} kg")
```

# Challenge 2: Crypto Market Analysis

## Goal

Analyze cryptocurrency prices using real-time data.

## Tasks

1. Get prices for:
   - bitcoin
   - ethereum
   - solana
   - dogecoin

2. Convert the response into a DataFrame (coins as rows)

3. Rename columns properly

4. Add:
   - price_rank (ranking by price)
   - price_category:
        - "high" if > 1000  
        - "medium" if between 100 and 1000  
        - "low" if < 100  

5. Answer:
   - Which coin is the most expensive?
   - Which category has more coins?

6. **Bonus**:
   - Create a column with price difference vs bitcoin

```python
# starter code
import os
import requests
import pandas as pd

url = "https://api.coingecko.com/api/v3/simple/price"

params = {"ids": "bitcoin,ethereum,solana,dogecoin","vs_currencies":"usd",}


response = requests.get(url,params=params, timeout =10)

response.raise_for_status()
price_data = response.json()
print(price_data)
```

```python
"""Convert the response into a DataFrame (coins as rows)"""
df_crypto = pd.DataFrame(price_data).T
df_crypto
```

```python
"""Rename columns properly"""
df_crypto = (pd.DataFrame.from_dict(price_data, orient="index").reset_index(names="coins").rename(columns={"usd":"price_in_usd"}))
df_crypto
```

```python
"""price_rank (ranking by price)"""


df_crypto=df_crypto.sort_values(by=["price_in_usd"], ascending=False)
display(df_crypto)


            

```

```python
"""price_category:
"high" if > 1000
"medium" if between 100 and 1000
"low" if < 100"""

import numpy as np
conditions = [df_crypto["price_in_usd"]>1000,(df_crypto["price_in_usd"]>=100)&(df_crypto["price_in_usd"]<=1000),df_crypto["price_in_usd"]<100]
values =["high","medium","low"]
df_crypto["price_category"] = np.select(conditions, values, default="unknown")
df_crypto



```

```python
"""Which coin is the most expensive?"""
most_expensive =df_crypto.loc[df_crypto["price_in_usd"].idxmax(),"coins"]
display(f"the most expensive coin is {most_expensive}")
```

```python
"""Which category has more coins?"""
category_more_coins=df_crypto["price_category"].value_counts().idxmax()
display(f"the category who has the more coins is {category_more_coins}")
```
