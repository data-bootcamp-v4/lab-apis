# Exercise 1: Top Tracks and Related Artists

```python
# get top tracks funtion 
def get_top_tracks(artist_name, top_n=5):
    res = sp.search(q=artist_name, limit=1, type='artist')
    artist_id = res['artists']['items'][0]['id']
    top_tracks = sp.artist_top_tracks(artist_id)['tracks']
    return [track['name'] for track in top_tracks[:top_n]]

# Example usage
artists = ["Lady Gaga", "Ed Sheeran", "Dua Lipa"]
top_tracks_list = [get_top_tracks(a) for a in artists]
print(top_tracks_list)

# find related artists function 
from spotipy.exceptions import SpotifyException

def find_related_artists(artist_name, top_n=5):
    try:
        res = sp.search(q=artist_name, limit=1, type='artist')
        artist_id = res['artists']['items'][0]['id']
        related_artists = sp.artist_related_artists(artist_id)['artists']
        return [artist['name'] for artist in related_artists[:top_n]]
    except (IndexError, SpotifyException):
        print(f"No related artists found for {artist_name}")
        return []

# Example usage
artists = ["Lady Gaga", "Ed Sheeran", "Dua Lipa"]
related_artists_list = [find_related_artists(a) for a in artists]
print(related_artists_list)

# combine for playlist
playlist_tracks = []

# Add top tracks of main artists
for artist in artists:
    playlist_tracks.extend(get_top_tracks(artist))

# Add top tracks of related artists
for artist in related_artists_list:
    for related_artist in artist:
        playlist_tracks.extend(get_top_tracks(related_artist))

print("Final playlist tracks:")
print(playlist_tracks[:20])  # Show first 20 tracks
```

# Exercise 2: Featured Playlists

```python
# Search for playlists

search_results = sp.search(q="Top Hits", type="playlist", limit=5)

# Safely get playlist items
featured = search_results.get('playlists', {}).get('items', [])

# Remove None entries
featured = [pl for pl in featured if pl is not None]

if not featured:
    print("No playlists found.")
else:
    # List the top playlists
    for i, playlist in enumerate(featured, 1):
        name = playlist.get('name', 'Unknown')
        pid = playlist.get('id', 'Unknown')
        print(f"{i}. {name} (ID: {pid})")

    # Choose one playlist (first one)
    playlist_id = featured[0].get('id')
    if playlist_id:
        playlist_info = sp.playlist(playlist_id)
        print("Name:", playlist_info.get('name', 'Unknown'))
        print("Description:", playlist_info.get('description', 'No description'))
        print("Total tracks:", playlist_info.get('tracks', {}).get('total', 0))

        # Extract first 10 tracks
        tracks = sp.playlist_tracks(playlist_id).get('items', [])[:10]
        track_artist_dict = {}
        for track_item in tracks:
            track = track_item.get('track', {})
            track_name = track.get('name', 'Unknown')
            artists = [artist.get('name', 'Unknown') for artist in track.get('artists', [])]
            track_artist_dict[track_name] = artists

        print(track_artist_dict)

```
